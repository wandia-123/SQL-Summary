# Database normalization
It is used to remove duplicates and ensure consistency.

**Database anomalies** - problems that occur in a poorly organized database(repetitive data) making it difficult to insert, delete or update data without causing inconsistency

**Types of anomalies**


1.   Insertion Anomaly
2.   Deletion Anomaly
3.   Update Anomaly

**Levels of Normalization**

They are called *Normal Forms*
Normal          Form	Main requirement

1. 1NF	        Each cell contains one value; no repeating groups.

2. 2NF	        1NF + No partial dependencies.

3. 3NF	        2NF + No transitive dependencies.

4. BCNF      	  Every determinant must be a candidate key.

*Partial Dependancy - An attribute depends only on one part of a composite key*
*Transitive Dependancy - An attribute depends on a non-primary attribute key that depends on a primary key*

# **SQL JOIN**
An SQL join is an operation used to combine data from two or more tables based on a related column between them.

**TYPES OF JOINS**


1. *Inner Join:-*  Returns only records that have matching values in both tables.



2. *Left Join-:*  Returns all records from the left table and matching records from the right table.



3. *Right Join-:* all records from the right table and matching records from the left table.



4. *Full Outer Join-:* Returns all records from both tables, including matching and non-matching records

**Important JOIN terms**

Join--Combines data from tables.

On--Specifies the condition used to match records.

Primary Key--Uniquely identifies each record in a table.

Foreign Key--Links one table to another.

Inner Join--Returns matching records only.

# ***SQL Queries with Expressions***



*   An expression is a combination of values, column names, operators and functions used to calculate or produce a new value in a query.

# ***AGGREGATE FUNCTIONS***

---

Group by is used to determine how records should be separated before functions performed by aggregate functions.
 Basic Syntax:

 *SELECT column_name, AGG_FUNCTION(column) FROM table GROUP BY column_name;*

 i.e: SELECT role, AVG(years_employed) FROM employees GROUP BY role;

 Step 1: Identify the unique roles-

* Group by can be used together with each of the aggregates studied above.



* GROUP BY can also be used together with AS to give selected columns meaningful names

E.g.

*SELECT role, AVG(years_employed) AS average_years
FROM employees

Important rules when using GROUP BY


**Rule 1: Every non-aggregated column in SELECT should normally appear in GROUP BY.**

Correct:
SELECT role, COUNT(*)
FROM employees
GROUP BY role;


Incorrect:
SELECT role, name, COUNT(*)
FROM employees
GROUP BY role;


**Rule 2: GROUP BY can be used with multiple aggregate functions.**

example:
SELECT role,
       COUNT(*) AS total_employees,
       AVG(years_employed) AS average_years,
       MAX(years_employed) AS longest_service
FROM employees
GROUP BY role;

**Rule 3: GROUP BY does not necessarily sort the results.**

If you want the results arranged in a particular order, use ORDER BY.

SELECT role, COUNT(*) AS total
FROM employees
GROUP BY role
ORDER BY total DESC;

# ***HAVING***

Having is used to filter grouped records based on the specified conditions ussualy involving aggregate functions.

* WHERE filters individual rows.

* GROUP BY combines similar rows into groups.

* HAVING filters the resulting groups.

Basic Syntax:

*SELECT group_by_column, AGG_FUNC(column_expression) AS aggregate_result_alias,
FROM mytable
WHERE condition
GROUP BY column
HAVING group_condition;*

Example: Understanding the problem

Imagine we have the Employees table with three roles:

Engineer	5,
Artist	5,
Manager	3

Suppose someone asks:

Find roles that have more than 3 employees.

We first need to:
1. Group employees according to their roles.
2. Count employees in each role.
3. Keep only roles whose employee count is greater than 3.

This is where HAVING comes in.

* Having can be used with all aggregates.

eg. in the above example with count

SELECT role, COUNT(*) FROM employees GROUP BY role HAVING COUNT(*) > 3;

**WHERE vs. HAVING**

Both filter data but at different stages.


| WHERE | HAVING
|----------|----------
| Filter individual rows  | Filter groups
| Applied before grouping | Applied after grouping
|Ussualy works with ordinary column conditions | Commonly works with aggregate conditions
|Can not directly use aggregate function in its conditions | Can use aggregate funtions in its conditions

* We can also use having with multiple conditions. This is achieved by using AND and OR.

Example:

SELECT role, AVG(years_employed)
FROM employees
GROUP BY role
HAVING AVG(years_employed) < 4
OR AVG(years_employed) > 7;

* Having can also be used with an alias(AS).

Example:

SELECT role, AVG(years_employed) AS average_years
FROM employees
GROUP BY role
HAVING average_years > 5;

***Overall Summary***

| Clause | Short meaning
| ---------- | -----------
| WHERE | Filters individual rows before grouping
| GROUP BY | Groups rows with similar values
| HAVING | Filters groups after aggregation
| ORDER BY | Sort the results
| LIMITS | Restricts the number of results

# ***ORDER OF EXECUTION OF A QUERY***

- Order of execution of a query refers to the sequence in which the database processes the different clauses of a query.
  


They are executed in the following order:

| Order Clause | Purpose |
|---|---|
| FROM | Identifies the table(s) to retrieve data from |
| JOIN | Combines data from multiple tables |
| ON | Specifies the joining condition |
| WHERE | Filters individual rows |
| GROUP BY | Groups rows with similar values |
| HAVING | Filters the grouped results |
| SELECT | Selects the column expressions to display |
| DISTINCT | Removes duplicate results |
| ORDER BY | Sorts the results |
| LIMIT | Restricts the number of rows displayed |
| OFFSET | Skips a specified number of rows |

*Complete select query*

 SELECT DISTINCT column, AGG_FUNC(column_or_expression), …

 FROM mytable

    JOIN another_table

      ON mytable.column = another_table.column

    WHERE constraint_expression

    GROUP BY column

    HAVING constraint_expression

    ORDER BY column ASC/DESC

    LIMIT count OFFSET COUNT;

#***Inserting Rows***

The INSERT INTO statement is used to add new records (rows) into an existing database table.

Basic syntax

**1. Inserting values into all columns**

*INSERT INTO table_name
VALUES (value1, value2, value3); *


- Values must be provided for all columns.
- Values must follow the order of the table columns.

**2. Inserting into specific columns**

*INSERT INTO table_name (column1, column2)
VALUES (value1, value2);*

- Specifies the columns receiving data.
- Columns not specified may receive default values or NULL, if allowed.
- Useful when a table has an AUTO_INCREMENT ID column.

 **3. Inserting multiple rows**


*Insert several records using one statement.
INSERT INTO table_name (column1, column2)
VALUES
    (value1, value2),
    (value3, value4),
    (value5, value6);*

Each set of parentheses represents one new row.

*Example*

INSERT INTO movies (title, director, year, length_minutes)
VALUES
    ('Toy Story 4', 'Josh Cooley', 2019, 100),
    ('Frozen', 'Chris Buck', 2013, 102);


This inserts two movie records into the movies table.

# ***Updating rows***

Update statement is used to modify existing records in a database.

Basic Syntax

***1. Updating Multiple columns***

It is possible to update more than one column in the same SQL querry.

Syntax:

*UPDATE table_name SET column1_name = new_value, column2_name = new_value WHERE condition....;*

Example

UPDATE students
SET name = 'Alice Wanjiku',
    course = 'Computer Science'
WHERE student_id = 1;

* Always check the where condition before executing an UPDATE
* Omitting where updates every row.

Expressions allows you to perform actions directely when updating a column.
